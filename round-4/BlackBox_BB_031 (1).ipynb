```python
# ============================================================
# BLACK BOX AI — LENDING ELIGIBILITY EMULATOR
# Team: BB-031
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PolynomialFeatures
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score

# ============================================================
# 1. DATA
# ============================================================

data = [
# score, age, credit_history, debt_ratio, dependents,
# employment_years, income, loan_amount, num_accounts,
# recent_defaults, region

[0.8900,40,452.5,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,452,0.7,5,1,10,50,17,5,'A'],
[0.8896,40,453,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,451,0.7,5,1,10,50,17,5,'A'],
[0.8900,40,455,0.7,5,1,10,50,17,5,'A'],
[0.8890,40,460,0.7,5,1,10,50,17,5,'A'],
[0.8891,40,475,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,450,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,425,0.7,5,1,10,50,17,5,'A'],
[0.8765,40,600,0.7,5,1,10,50,17,5,'A'],
[0.8867,40,500,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,400,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,350,0.7,5,1,10,50,17,5,'A'],

[0.0200,50,300.559,0.7,5,1,10,50,17,5,'A'],
[0.0200,45,300.559,0.7,5,1,10,50,17,5,'A'],
[0.0200,40,300.559,0.7,5,1,10,50,17,5,'A'],
[0.0200,35,300.559,0.7,5,1,10,50,17,5,'A'],
[0.0200,30,300.559,0.7,5,1,10,50,17,5,'A'],
[0.0200,25,300.559,0.7,5,1,10,50,17,5,'A'],
[0.8548,40,500,0.7,5,1,10,20,17,5,'A'],
[0.9501,75,800,0.7,5,1,10,70,17,5,'A'],
[0.9199,60,900,0.7,5,1,10,50,17,5,'A'],
[0.0200,20,300.559,0.7,5,1,10,50,17,5,'A'],

[0.8581,30,300.559,0.7,5,1,10,70,17,5,'D'],
[0.8664,30,300.559,0.7,5,1,10,70,17,5,'C'],
[0.8710,30,300.559,0.7,5,1,10,70,17,5,'B'],
[0.0200,30,300.559,0.7,5,1,10,70,17,5,'A'],

[0.9671,75,300.559,0.7,5,1,10,70,17,5,'D'],
[0.9670,75,300.559,0.7,5,1,10,70,17,5,'C'],
[0.9674,75,300.559,0.7,5,1,10,70,17,5,'B'],
[0.0200,75,300.559,0.7,5,1,10,70,17,5,'A'],

[0.9208,60,900,0.7,5,1,10,50,17,5,'C'],
[0.9262,60,850,0.7,5,1,10,50,17,5,'C'],
[0.9276,60,800,0.7,5,1,10,50,17,5,'C'],
[0.9333,60,750,0.7,5,1,10,50,17,5,'C'],
[0.9368,60,700,0.7,5,1,10,50,17,5,'C'],
[0.9357,60,650,0.7,5,1,10,50,17,5,'C'],
[0.9407,60,600,0.7,5,1,10,50,17,5,'C'],
[0.9440,60,550,0.7,5,1,10,50,17,5,'C'],
[0.9443,60,500,0.7,5,1,10,50,17,5,'C'],
[0.9489,60,450,0.7,5,1,10,50,17,5,'C'],
[0.9553,60,400,0.7,5,1,10,50,17,5,'C'],
[0.9526,60,350,0.7,5,1,10,50,17,5,'C'],
[0.9498,60,300,0.7,5,1,10,50,17,5,'C'],

[0.9670,75,300.559,0.7,5,1,10,70,17,5,'C'],
[0.9668,75,300.559,0.7,5,1,10,65,17,5,'C'],
[0.9661,75,300.559,0.7,5,1,10,60,17,5,'C'],
[0.9666,75,300.559,0.7,5,1,10,55,17,5,'C'],
[0.9673,75,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9663,75,300.559,0.7,5,1,10,45,17,5,'C'],
[0.9652,75,300.559,0.7,5,1,10,40,17,5,'C'],
[0.9626,75,300.559,0.7,5,1,10,30,17,5,'C'],
[0.9618,75,300.559,0.7,5,1,10,25,17,5,'C'],
[0.9584,75,300.559,0.7,5,1,10,20,17,5,'C'],
[0.9517,75,300.559,0.7,5,1,10,15,17,5,'C'],
[0.9465,75,300.559,0.7,5,1,10,10,17,5,'C'],
[0.9438,75,300.559,0.7,5,1,10,5,17,5,'C'],

[0.9655,72,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9621,70,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9599,65,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9498,60,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9349,55,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9268,50,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9137,45,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9009,40,300.559,0.7,5,1,10,50,17,5,'C'],
[0.8838,35,300.559,0.7,5,1,10,50,17,5,'C'],
[0.8655,30,300.559,0.7,5,1,10,50,17,5,'C'],
[0.8479,25,300.559,0.7,5,1,10,50,17,5,'C'],
[0.7995,20,300.559,0.7,5,1,10,50,17,5,'C'],
[0.9673,75,300.559,0.7,5,1,10,50,17,5,'C'],

# Round 2
[0.9512,75,811.559,1,6,3.5,10,100,17,5,'C'],
[0.8840,75,811.559,1,6,3.5,10,1,17,5,'C'],
[0.9648,75,300.559,1,6,3.5,10,100,17,5,'C'],
[0.9430,75,300.559,1,6,3.5,10,1,17,5,'C'],
[0.0200,46.5,300.559,.5,3,20,50,50,10,2.5,'A'],
[0.0200,46.5,301,.5,3,20,50,50,10,2.5,'A'],
[0.5528,46.5,900,.5,3,20,50,50,10,2.5,'A'],
[0.5931,46.5,811.559,.5,3,20,50,100,10,2.5,'A'],
[0.3975,46.5,811.559,.5,3,20,50,1,10,2.5,'A'],
[0.0200,46.5,300.559,.5,3,20,50,50,10,2.5,'A'],
[0.0200,46.5,300.559,.5,3,20,50,100,10,2.5,'A'],
[0.0200,46.5,300.559,.5,3,20,50,1,10,2.5,'A'],
[0.9487,75,811.559,1,6,3.5,10,50,17,5,'C'],
[0.9676,75,500.559,1,6,3.5,10,50,17,5,'C'],
[0.9707,75,300.1,1,6,3.5,10,50,17,5,'C'],
[0.9707,75,300,1,6,3.5,10,50,17,5,'C'],
[0.9648,75,300.559,1,6,3.5,10,100,17,5,'C'],
[0.9430,75,300.559,1,6,3.5,10,1,17,5,'C'],
[0.9707,75,300.559,1,6,3.5,10,50,17,5,'C'],

# Round 1
[0.9707,75,300.559,1,6,3.5,10,50,17,5,'C'],
[0.9699,75,300.559,1,6,3.9,10,50,17,5,'C'],
[0.9707,75,300.559,1,6,3.5,10,50,17,5,'C'],
[0.9699,75,300.559,1,6,4,10,50,17,5,'C'],
[0.9701,75,300.559,1,6,3,10,50,17,5,'C'],
[0.9661,75,300,1,6,2,100,50,18.5,5,'C'],
[0.9661,75,300.559,1,6,2,100,50,18.5,5,'C'],
[0.9661,75,300.559,1,6,2,0,50,18.5,5,'C'],
[0.9670,75,300.559,1,6,2,10,55,18.5,5,'C'],
[0.9672,75,300.559,1,6,2,10,55,15.5,5,'C'],
[0.9678,75,300.559,1,6,2,10,55,17,5,'C'],
[0.9673,75,300.559,1,6,2,10,60,17,5,'C'],
[0.9664,75,300.559,1,6,2,10,40,17,5,'C'],
[0.9681,75,300.559,1,6,2,10,50,17,5,'C'],
[0.9673,75,300.559,1,6,1,10,50,17,5,'C'],
[0.9673,75,300.559,1,0,1,10,50,17,5,'C'],
[0.9673,75,300.559,1,5,1,10,50,17,5,'C'],
[0.9673,75,0,5,1,10,50,17,5,'C'],
[0.9656,75,300.559,.7,5,0,10,50,17,5,'C'],
[0.9673,75,300.559,.7,5,1,10,50,17,5,'C'],
[0.9492,75,300.559,.7,5,1,10,50,17,3,'C'],
[0.8737,75,300.559,.7,5,1,10,50,17,0,'C'],
[0.9166,75,300.559,.7,5,1,10,50,17,1.298,'D'],
[0.9133,75,300.559,.7,5,1,10,50,17,1.298,'B'],
[0.0200,75,300.559,.7,5,1,10,50,17,1.298,'A'],
[0.9209,75,300.559,.7,5,1,10,50,17,1.298,'C'],
[0.9002,75,300.559,.7,5,1,10,30,17,1.298,'C'],
[0.9186,75,300.559,.7,5,1,10,60,19,1.298,'C'],
[0.9202,75,300.559,.7,5,1,10,60,17,1.298,'C'],
[0.9209,75,300.559,.7,5,1,10,50,17,1.298,'C'],
[0.8722,75,300.559,.7,5,1,10,50,10,1.298,'C'],
[0.9187,75,300.559,.7,5,1,10,50,16.213,'C'],
[0.9122,75,300.559,.7,5,1,10,100,16.213,'C'],
[0.8627,75,300.559,.7,5,1,10,1,16.213,'C'],
[0.9169,75,300.559,.7,5,1,10,77.854,16.213,'C'],
[0.9169,75,300.559,.7,5,1,90,77.854,16.213,'C'],
[0.9169,75,300.559,.7,5,1,47.804,77.854,16.213,'C'],
[0.8353,75,300.559,.7,20,47.804,77.854,16.213,1.298,'C'],
[0.8851,75,300.559,.7,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8851,75,300.559,.7,3,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8851,75,300.559,.7,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8851,75,300.559,.5,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8769,75,500.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8851,75,300.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8444,75,811.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.8351,70.035,811.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.7940,61.035,811.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.5426,36.186,811.559,.167,2.964,12.407,47.804,77.854,16.213,1.298,'C'],
[0.0639,21.5,900,.1,0,40,99,10,1,1,'A'],
[0.6146,46.5,600,.5,3,20,50,50,10,2.5,'A'],
]

columns = [
    'score','age','credit_history','debt_ratio','dependents',
    'employment_years','income','loan_amount','num_accounts',
    'recent_defaults','region'
]

df = pd.DataFrame(data, columns=columns)

print("Dataset shape:", df.shape)
print(df.head())

# ============================================================
# 2. CLEAN DATA
# ============================================================

numeric_features = [
    'age','credit_history','debt_ratio','dependents',
    'employment_years','income','loan_amount',
    'num_accounts','recent_defaults'
]

X = df.drop(columns=['score'])
y = df['score']

# ============================================================
# 3. TRAIN / VALIDATION SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# ============================================================
# 4. PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), ['region'])
    ]
)

# ============================================================
# 5. MODEL 1 — EXTRA TREES
# ============================================================

model = Pipeline([
    ('prep', preprocessor),
    ('model', ExtraTreesRegressor(
        n_estimators=500,
        max_depth=None,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1
    ))
])

model.fit(X_train, y_train)

pred = np.clip(model.predict(X_test), 0, 1)

mae = mean_absolute_error(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))
r2 = r2_score(y_test, pred)

decision_true = (y_test >= 0.5).astype(int)
decision_pred = (pred >= 0.5).astype(int)

decision_acc = accuracy_score(decision_true, decision_pred)

print("\n================ MODEL RESULTS ================")
print(f"MAE              : {mae:.5f}")
print(f"RMSE             : {rmse:.5f}")
print(f"R²               : {r2:.5f}")
print(f"Decision Accuracy: {decision_acc:.2%}")

# ============================================================
# 6. OBSERVED REGION-A CONDITIONAL RULE
# ============================================================
#
# From controlled experiments:
# age=40, region=A, debt=.7, dependents=5,
# employment=1, income=10, loan=50,
# accounts=17, defaults=5
#
# credit 452   -> 0.0200
# credit 452.5 -> 0.8900
#
# This is an observed discontinuity.
# We only apply the rule in this exact tested configuration.

def observed_region_A_rule(row, predicted_score):

    if (
        row['region'] == 'A'
        and abs(row['age'] - 40) < 1e-9
        and abs(row['debt_ratio'] - 0.7) < 1e-9
        and row['dependents'] == 5
        and row['employment_years'] == 1
        and row['income'] == 10
        and row['loan_amount'] == 50
        and row['num_accounts'] == 17
        and row['recent_defaults'] == 5
    ):
        if row['credit_history'] <= 452:
            return 0.0200
        elif row['credit_history'] >= 452.5:
            return predicted_score

    return predicted_score


# ============================================================
# 7. FINAL PREDICT FUNCTION
# ============================================================

def predict(
    age,
    credit_history,
    debt_ratio,
    dependents,
    employment_years,
    income,
    loan_amount,
    num_accounts,
    recent_defaults,
    region
):

    row = pd.DataFrame([{
        'age': age,
        'credit_history': credit_history,
        'debt_ratio': debt_ratio,
        'dependents': dependents,
        'employment_years': employment_years,
        'income': income,
        'loan_amount': loan_amount,
        'num_accounts': num_accounts,
        'recent_defaults': recent_defaults,
        'region': region
    }])

    score = float(model.predict(row)[0])
    score = np.clip(score, 0, 1)

    score = observed_region_A_rule(row.iloc[0], score)

    decision = "APPROVE" if score >= 0.5 else "DECLINE"

    return round(float(score), 4), decision


# ============================================================
# 8. TEST EXAMPLES
# ============================================================

tests = [
    # Known decline
    [40,452,0.7,5,1,10,50,17,5,'A'],

    # Known approve
    [40,452.5,0.7,5,1,10,50,17,5,'A'],

    # Known approve
    [75,300.559,0.7,5,1,10,50,17,5,'C'],

    # Known decline
    [75,300.559,0.7,5,1,10,50,17,5,'A'],
]

print("\n================ TEST PREDICTIONS ================")

for t in tests:

    score, decision = predict(*t)

    print(
        f"score={score:.4f} | "
        f"decision={decision}"
    )


# ============================================================
# 9. FEATURE IMPORTANCE
# ============================================================

rf = model.named_steps['model']

feature_names = (
    numeric_features +
    list(model.named_steps['prep']
         .named_transformers_['cat']
         .get_feature_names_out(['region']))
)

importance = pd.Series(
    rf.feature_importances_,
    index=feature_names
).sort_values(ascending=False)

print("\n================ FEATURE IMPORTANCE ================")
print(importance)


# ============================================================
# 10. SAVE MODEL
# ============================================================

import joblib

joblib.dump(model, "black_box_lending_emulator.pkl")

print("\nModel saved as black_box_lending_emulator.pkl")
print("\nDONE — BB-031 emulator ready.")
```

```python
# ============================================================
# BB-031 — TRAIN ON R1 + R2, VALIDATE ON UNSEEN R4
# ============================================================

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np
import pandas as pd
import joblib


# ============================================================
# 1. FEATURES
# ============================================================

features = [
    'age',
    'credit_history',
    'debt_ratio',
    'dependents',
    'employment_years',
    'income',
    'loan_amount',
    'num_accounts',
    'recent_defaults',
    'region'
]

numeric_features = [
    'age',
    'credit_history',
    'debt_ratio',
    'dependents',
    'employment_years',
    'income',
    'loan_amount',
    'num_accounts',
    'recent_defaults'
]


# ============================================================
# 2. SEPARATE ROUNDS
# ============================================================

# Your source data is ordered:
#
# Rows 0-79   = Round 4 (80 queries)
# Rows 80-98  = Round 2 (19 queries)
# Rows 99-138 = Round 1 (40 queries)

round4 = df.iloc[:80].copy()
round2 = df.iloc[80:99].copy()
round1 = df.iloc[99:139].copy()

# Training = Round 1 + Round 2
train_df = pd.concat(
    [round1, round2],
    ignore_index=True
)

# Validation = Round 4 ONLY
test_df = round4.copy()


print("========================================")
print("          DATASET SPLIT")
print("========================================")
print(f"Round 1 samples       : {len(round1)}")
print(f"Round 2 samples       : {len(round2)}")
print(f"Training samples      : {len(train_df)}")
print(f"Round 4 samples       : {len(round4)}")
print(f"Unseen test samples   : {len(test_df)}")
print("========================================")


# ============================================================
# 3. PREPARE TRAINING DATA
# ============================================================

X_train = train_df[features].copy()
y_train = train_df['score'].copy()

X_test = test_df[features].copy()
y_test = test_df['score'].copy()

for col in numeric_features:
    X_train[col] = pd.to_numeric(
        X_train[col],
        errors='coerce'
    )

    X_test[col] = pd.to_numeric(
        X_test[col],
        errors='coerce'
    )

X_train['region'] = X_train['region'].astype(str)
X_test['region'] = X_test['region'].astype(str)


# ============================================================
# 4. PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer([
    (
        'numeric',
        'passthrough',
        numeric_features
    ),
    (
        'region',
        OneHotEncoder(handle_unknown='ignore'),
        ['region']
    )
])


# ============================================================
# 5. MODEL
# ============================================================

model = Pipeline([
    (
        'preprocessor',
        preprocessor
    ),
    (
        'model',
        ExtraTreesRegressor(
            n_estimators=500,
            random_state=42,
            max_features=1.0,
            min_samples_leaf=1,
            n_jobs=-1
        )
    )
])


# ============================================================
# 6. TRAIN
# ============================================================

# IMPORTANT:
# Round 4 is NOT used here.

model.fit(
    X_train,
    y_train
)

print("\nModel trained ONLY on Round 1 + Round 2.")


# ============================================================
# 7. PREDICT UNSEEN ROUND 4
# ============================================================

pred = model.predict(X_test)

pred = np.clip(
    pred,
    0,
    1
)


# ============================================================
# 8. SCORE METRICS
# ============================================================

mae = mean_absolute_error(
    y_test,
    pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        pred
    )
)

r2 = r2_score(
    y_test,
    pred
)


# ============================================================
# 9. DECISION ACCURACY
# ============================================================

actual_decision = (
    y_test >= 0.5
)

predicted_decision = (
    pred >= 0.5
)

decision_accuracy = np.mean(
    actual_decision == predicted_decision
)


# ============================================================
# 10. FINAL VALIDATION RESULTS
# ============================================================

print("\n")
print("========================================")
print("     ROUND 4 UNSEEN VALIDATION")
print("========================================")
print(f"Training samples   : {len(X_train)}")
print(f"Unseen samples     : {len(X_test)}")
print("----------------------------------------")
print(f"MAE                : {mae:.6f}")
print(f"RMSE               : {rmse:.6f}")
print(f"R²                 : {r2:.6f}")
print(f"Decision Accuracy  : {decision_accuracy:.2%}")
print("========================================")


# ============================================================
# 11. DETAILED RESULTS
# ============================================================

results = test_df.copy()

results['predicted_score'] = pred

results['absolute_error'] = (
    results['score'] -
    results['predicted_score']
).abs()

results['actual_decision'] = np.where(
    results['score'] >= 0.5,
    'APPROVE',
    'DECLINE'
)

results['predicted_decision'] = np.where(
    results['predicted_score'] >= 0.5,
    'APPROVE',
    'DECLINE'
)

results['decision_correct'] = (
    results['actual_decision'] ==
    results['predicted_decision']
)


# ============================================================
# 12. WORST SCORE PREDICTIONS
# ============================================================

print("\nTOP 15 SCORE PREDICTION ERRORS")

display(
    results.sort_values(
        'absolute_error',
        ascending=False
    )[
        [
            'age',
            'credit_history',
            'loan_amount',
            'region',
            'score',
            'predicted_score',
            'absolute_error',
            'actual_decision',
            'predicted_decision',
            'decision_correct'
        ]
    ].head(15)
)


# ============================================================
# 13. WRONG DECISIONS
# ============================================================

wrong = results[
    results['decision_correct'] == False
]

print("\nWRONG DECISIONS:", len(wrong))

if len(wrong) > 0:

    display(
        wrong[
            [
                'age',
                'credit_history',
                'loan_amount',
                'region',
                'score',
                'predicted_score',
                'absolute_error',
                'actual_decision',
                'predicted_decision'
            ]
        ]
    )

else:

    print("No wrong decisions on unseen Round 4.")


# ============================================================
# 14. FINAL PREDICTION FUNCTION
# ============================================================

def predict(
    age,
    credit_history,
    debt_ratio,
    dependents,
    employment_years,
    income,
    loan_amount,
    num_accounts,
    recent_defaults,
    region
):

    applicant = pd.DataFrame([{
        'age': age,
        'credit_history': credit_history,
        'debt_ratio': debt_ratio,
        'dependents': dependents,
        'employment_years': employment_years,
        'income': income,
        'loan_amount': loan_amount,
        'num_accounts': num_accounts,
        'recent_defaults': recent_defaults,
        'region': str(region)
    }])

    score = float(
        model.predict(applicant)[0]
    )

    score = np.clip(
        score,
        0,
        1
    )

    decision = (
        "APPROVE"
        if score >= 0.5
        else "DECLINE"
    )

    return round(float(score), 4), decision


# ============================================================
# 15. KNOWN CASES
# ============================================================

print("\nKNOWN CASE TESTS")

tests = [
    [40, 452, 0.7, 5, 1, 10, 50, 17, 5, 'A'],
    [40, 452.5, 0.7, 5, 1, 10, 50, 17, 5, 'A'],
    [75, 300.559, 0.7, 5, 1, 10, 50, 17, 5, 'C'],
    [75, 300.559, 0.7, 5, 1, 10, 50, 17, 5, 'A']
]

for x in tests:

    score, decision = predict(*x)

    print(
        f"credit={x[1]}, "
        f"region={x[-1]} "
        f"--> score={score}, "
        f"{decision}"
    )


# ============================================================
# 16. SAVE MODEL
# ============================================================

joblib.dump(
    model,
    "BB031_R1_R2_TRAINED_R4_VALIDATED.pkl"
)

print(
    "\nModel saved:"
    " BB031_R1_R2_TRAINED_R4_VALIDATED.pkl"
)
```

```python
# ============================================================
# BB-031 FINAL HYBRID EMULATOR
# CLEAN + TRAIN
# ============================================================

import numpy as np
import pandas as pd
import joblib

from sklearn.ensemble import ExtraTreesRegressor


# ============================================================
# 1. FEATURES
# ============================================================

features = [
    "age",
    "credit_history",
    "debt_ratio",
    "dependents",
    "employment_years",
    "income",
    "loan_amount",
    "num_accounts",
    "recent_defaults",
    "region"
]


# ============================================================
# 2. CREATE FRESH COPY
# ============================================================

data_final = df[features + ["score"]].copy()


# ============================================================
# 3. FIND VALID REGIONS
# ============================================================

valid_regions = ["A", "B", "C", "D"]

valid_region_mask = data_final["region"].isin(valid_regions)

print("============================================================")
print("DATA QUALITY CHECK")
print("============================================================")
print("Total rows:", len(data_final))
print("Valid region rows:", valid_region_mask.sum())
print("Invalid/missing region rows:", (~valid_region_mask).sum())


# ============================================================
# 4. DISPLAY MALFORMED ROWS
# ============================================================

bad_rows = data_final[~valid_region_mask]

if len(bad_rows) > 0:

    print("\n============================================================")
    print("MALFORMED / INCOMPLETE ROWS")
    print("============================================================")

    print(
        bad_rows[
            [
                "age",
                "credit_history",
                "debt_ratio",
                "dependents",
                "employment_years",
                "income",
                "loan_amount",
                "num_accounts",
                "recent_defaults",
                "region",
                "score"
            ]
        ].to_string()
    )


# ============================================================
# 5. TRAIN ONLY ON VALID OBSERVATIONS
# ============================================================

clean_df = data_final[valid_region_mask].copy()

print("\n============================================================")
print("CLEAN TRAINING DATA")
print("============================================================")
print("Rows used for training:", len(clean_df))


# ============================================================
# 6. ENCODE REGION
# ============================================================

region_mapping = {
    "A": 0,
    "B": 1,
    "C": 2,
    "D": 3
}

X_final = clean_df[features].copy()
y_final = clean_df["score"].astype(float).copy()

X_final["region"] = (
    X_final["region"]
    .astype(str)
    .str.strip()
    .map(region_mapping)
)


# Safety check
if X_final["region"].isna().any():

    print("ERROR: Region encoding failed.")
    print(X_final["region"].unique())

    raise ValueError("Region encoding failed")


# Make numeric
for col in features:
    X_final[col] = pd.to_numeric(
        X_final[col],
        errors="coerce"
    )


# Check for any remaining missing values
missing_counts = X_final.isna().sum()

if missing_counts.sum() > 0:

    print("\nMissing values detected:")
    print(missing_counts[missing_counts > 0])

    raise ValueError(
        "Training data still contains missing values."
    )


# ============================================================
# 7. TRAIN FINAL MODEL
# ============================================================

final_model = ExtraTreesRegressor(
    n_estimators=500,
    random_state=42,
    max_features=1.0,
    min_samples_leaf=1,
    n_jobs=-1
)

final_model.fit(X_final, y_final)


print("\n============================================================")
print("FINAL MODEL TRAINED")
print("============================================================")
print("Training rows:", len(X_final))
print("Trees:", 500)
print("Model: ExtraTreesRegressor")


# ============================================================
# 8. PREDICTION FUNCTION
# ============================================================

def predict_score(
    age,
    credit_history,
    debt_ratio,
    dependents,
    employment_years,
    income,
    loan_amount,
    num_accounts,
    recent_defaults,
    region
):

    # --------------------------------------------------------
    # DISCOVERED REGION-A THRESHOLD
    # --------------------------------------------------------

    exact_region_a_config = (
        region == "A"
        and age == 40
        and debt_ratio == 0.7
        and dependents == 5
        and employment_years == 1
        and income == 10
        and loan_amount == 50
        and num_accounts == 17
        and recent_defaults == 5
    )

    if exact_region_a_config:

        # Observed experimentally:
        #
        # credit <= 452.0 -> 0.0200
        # credit >= 452.5 -> APPROVE region

        if credit_history <= 452:
            return 0.0200


    # --------------------------------------------------------
    # NORMAL ML PREDICTION
    # --------------------------------------------------------

    encoded_region = region_mapping.get(region)

    if encoded_region is None:
        raise ValueError(
            f"Unknown region: {region}"
        )

    input_data = pd.DataFrame([{
        "age": float(age),
        "credit_history": float(credit_history),
        "debt_ratio": float(debt_ratio),
        "dependents": float(dependents),
        "employment_years": float(employment_years),
        "income": float(income),
        "loan_amount": float(loan_amount),
        "num_accounts": float(num_accounts),
        "recent_defaults": float(recent_defaults),
        "region": float(encoded_region)
    }])

    score = final_model.predict(input_data)[0]

    return float(
        np.clip(score, 0.0, 1.0)
    )


# ============================================================
# 9. DECISION FUNCTION
# ============================================================

def get_decision(score):

    return (
        "APPROVE"
        if score >= 0.5
        else "DECLINE"
    )


# ============================================================
# 10. COMPLETE PREDICT FUNCTION
# ============================================================

def predict(
    age,
    credit_history,
    debt_ratio,
    dependents,
    employment_years,
    income,
    loan_amount,
    num_accounts,
    recent_defaults,
    region
):

    score = predict_score(
        age,
        credit_history,
        debt_ratio,
        dependents,
        employment_years,
        income,
        loan_amount,
        num_accounts,
        recent_defaults,
        region
    )

    return {
        "score": round(score, 4),
        "decision": get_decision(score)
    }


# ============================================================
# 11. TEST KNOWN CASES
# ============================================================

print("\n============================================================")
print("KNOWN DISCOVERED CASES")
print("============================================================")


test_cases = [

    {
        "name": "Region A / Credit 452",
        "age": 40,
        "credit_history": 452,
        "debt_ratio": 0.7,
        "dependents": 5,
        "employment_years": 1,
        "income": 10,
        "loan_amount": 50,
        "num_accounts": 17,
        "recent_defaults": 5,
        "region": "A",
        "expected": 0.0200
    },

    {
        "name": "Region A / Credit 452.5",
        "age": 40,
        "credit_history": 452.5,
        "debt_ratio": 0.7,
        "dependents": 5,
        "employment_years": 1,
        "income": 10,
        "loan_amount": 50,
        "num_accounts": 17,
        "recent_defaults": 5,
        "region": "A",
        "expected": 0.8900
    },

    {
        "name": "Region C / Credit 300.559",
        "age": 75,
        "credit_history": 300.559,
        "debt_ratio": 0.7,
        "dependents": 5,
        "employment_years": 1,
        "income": 10,
        "loan_amount": 70,
        "num_accounts": 17,
        "recent_defaults": 5,
        "region": "C",
        "expected": 0.9673
    },

    {
        "name": "Region A / Credit 300.559",
        "age": 75,
        "credit_history": 300.559,
        "debt_ratio": 0.7,
        "dependents": 5,
        "employment_years": 1,
        "income": 10,
        "loan_amount": 70,
        "num_accounts": 17,
        "recent_defaults": 5,
        "region": "A",
        "expected": 0.0200
    }
]


for case in test_cases:

    result = predict(
        age=case["age"],
        credit_history=case["credit_history"],
        debt_ratio=case["debt_ratio"],
        dependents=case["dependents"],
        employment_years=case["employment_years"],
        income=case["income"],
        loan_amount=case["loan_amount"],
        num_accounts=case["num_accounts"],
        recent_defaults=case["recent_defaults"],
        region=case["region"]
    )

    print("\n" + case["name"])
    print("Expected :", case["expected"])
    print("Predicted:", result["score"])
    print("Decision :", result["decision"])


# ============================================================
# 12. SAVE FINAL MODEL
# ============================================================

model_package = {
    "model": final_model,
    "features": features,
    "region_mapping": region_mapping,
    "decision_threshold": 0.5,
    "training_rows": len(X_final),

    "discovered_region_A_rule": {
        "age": 40,
        "debt_ratio": 0.7,
        "dependents": 5,
        "employment_years": 1,
        "income": 10,
        "loan_amount": 50,
        "num_accounts": 17,
        "recent_defaults": 5,
        "credit_le_452": 0.0200,
        "credit_ge_452_5_observed": 0.8900
    }
}


MODEL_PATH = "BB031_FINAL_HYBRID_EMULATOR.pkl"

joblib.dump(
    model_package,
    MODEL_PATH
)


print("\n============================================================")
print("FINAL MODEL SAVED")
print("============================================================")
print(MODEL_PATH)
```
